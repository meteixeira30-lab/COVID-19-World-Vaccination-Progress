```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Usando o caminho exato que você acabou de listar
caminho = '/kaggle/input/datasets/gpreda/covid-world-vaccination-progress/country_vaccinations.csv'

# 2. Carregar o DataFrame
df = pd.read_csv(caminho)

# 3. Preparação essencial (Saúde Pública)
df['date'] = pd.to_datetime(df['date'])
df = df.fillna(0)

print("✅ Sucesso! Dados reais de vacinação carregados.")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filtrar o dataset para os países que você quer comparar
paises_analise = ['Brazil', 'United States', 'United Kingdom']
df_paises = df[df['country'].isin(paises_analise)]

# Criar o gráfico de linha para ver a velocidade de cobertura por milhão
plt.figure(figsize=(14, 7))
sns.lineplot(
    data=df_paises, 
    x='date', 
    y='daily_vaccinations_per_million', 
    hue='country',
    linewidth=2.5
)

plt.title('Velocidade de Resposta Vacinal: Doses Diárias Aplicadas por Milhão de Habitantes', fontsize=14, pad=15)
plt.xlabel('Linha do Tempo', fontsize=12)
plt.ylabel('Vacinações Diárias (Por Milhão)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Países')
plt.tight_layout()
plt.show()
```

```python
# Filtrar apenas Brasil
df_brasil = df[df['country'] == 'Brazil'].copy()

# Criar o gráfico
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_brasil, x='date', y='daily_vaccinations', color='green')
plt.title('Progresso da Vacinação no Brasil (Kaggle Dataset)')
plt.show()
```

 Ao analisar a curva de vacinação, observamos um crescimento acelerado a partir de meados de 2021. O ponto mais alto da curva indica o sucesso das campanhas de vacinação em massa. Comparando com o projeto de No-Show (onde usei dados sintéticos para prever faltas), aqui os dados reais confirmam que a adesão da população brasileira à vacina foi crescente e sustentada ao longo dos meses. Do ponto de vista epidemiológico, os picos representam a eficácia das campanhas de mobilização e a chegada de novos lotes.

```python
# 1. Selecionar os países para comparar
paises_comparacao = ['Brazil', 'Argentina', 'Chile', 'Colombia']
df_latam = df[df['country'].isin(paises_comparacao)].copy()

# 2. Criar o gráfico comparativo
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_latam, x='date', y='daily_vaccinations_per_million', hue='country', linewidth=2)

# 3. Customização profissional
plt.title('Vacinações Diárias por Milhão de Habitantes (Comparativo LATAM)', fontsize=15, fontweight='bold')
plt.xlabel('Data')
plt.ylabel('Vacinações por Milhão')
plt.legend(title='País')
plt.show()
```

Ao normalizarmos os dados para "vacinações por milhão", conseguimos uma comparação mais equiparada entre os países de diferentes populações. Observamos que nos países com picos mais acentuados tivemos uma capacidade de resposta mais rápida. Com base nisso, podemos em primeira análise  entender como diferentes políticas de saúde pública impactam a imunidade de grupo. A interlocucao que encontramos entre esses dados epidemiologicos e a analise de dados é justamente poder observar a velocidade de cobertura vacinal e o número total de doses ministradas pois dita a velocidade de recuperação funcional da sociedade.

```python
# 1. Agrupar por país e pegar o máximo de vacinações totais
top_10 = df.groupby('country')['total_vaccinations'].max().sort_values(ascending=False).head(10).reset_index()

# 2. Criar gráfico de barras
plt.figure(figsize=(12, 6))
sns.barplot(data=top_10, x='total_vaccinations', y='country', palette='magma')

# 3. Ajustar títulos
plt.title('Top 10 Países com Maior Número Total de Vacinas Aplicadas', fontsize=15, fontweight='bold')
plt.xlabel('Total de Vacinações (em Bilhões)')
plt.ylabel('País')
plt.show()
```

```python
# 1. Carregar o segundo arquivo (Caminho que você encontrou)
caminho_fab = '/kaggle/input/datasets/gpreda/covid-world-vaccination-progress/country_vaccinations_by_manufacturer.csv'
df_fab = pd.read_csv(caminho_fab)

# 2. Somar o total por vacina
ranking_vacinas = df_fab.groupby('vaccine')['total_vaccinations'].sum().sort_values(ascending=False).reset_index()

# 3. Criar o gráfico
plt.figure(figsize=(10, 6))
sns.barplot(data=ranking_vacinas, x='total_vaccinations', y='vaccine', palette='magma')
plt.title('Distribuição Global por Fabricante de Vacina', fontsize=15)
plt.show()
```

Ao analisar a distribuição das vacinas aplicadas, observamos uma predominância da Pfizer/BioNTech, seguida por outras farmacêuticas como Moderna e Oxford/AstraZeneca. Este gráfico revela a predominância da tecnologia de mRNA (como Pfizer) no cenário global. A dependência de fabricantes específicos levanta questões sobre a segurança da cadeia de suprimentos. Do ponto de vista de Epidemiologia, diferentes fabricantes possuem protocolos de dosagem distintos, o que exige uma gestão de dados rigorosa para garantir que os esquemas vacinais sejam completados corretamente, garantindo a eficácia clínica esperada.

```python
import warnings
warnings.filterwarnings('ignore')
# 1. Criar colunas auxiliares para tempo (Tratando os dados reais)
df_brasil['dia_semana'] = df_brasil['date'].dt.day_name()
df_brasil['mes'] = df_brasil['date'].dt.month_name()

# 2. Ordenar os dias para o gráfico ficar organizado
dias_ordenados = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df_brasil['dia_semana'] = pd.Categorical(df_brasil['dia_semana'], categories=dias_ordenados, ordered=True)

# 3. Criar a tabela dinâmica (Pivot Table) para o Heatmap
# Vamos calcular a média de vacinações por dia da semana em cada mês
heatmap_data = df_brasil.pivot_table(index='dia_semana', columns='mes', values='daily_vaccinations', aggfunc='mean')

# 4. Desenhar o Mapa de Calor
plt.figure(figsize=(14, 8))
sns.heatmap(heatmap_data, annot=True, fmt=".0f", cmap="YlOrRd")

plt.title('Mapa de Calor: Ritmo de Vacinação no Brasil\n(Média Diária por Dia da Semana e Mês)', fontsize=16, fontweight='bold')
plt.xlabel('Mês', fontsize=12)
plt.ylabel('Dia da Semana', fontsize=12)

plt.show()
```

O Mapa de Calor é uma ferramenta poderosa para identificar gargalos logísticos. Notei que o sistema de saúde brasileiro manteve uma operação resiliente. Se observarmos quedas sistemáticas aos fins de semana, identificamos barreiras de acesso. Em uma ánalise do ponto de vista ocupacional o acesso facilitado em diferentes dias da semana é um determinante social de saúde; se um paciente com dificuldades de mobilidade só pode ser levado por familiares ao domingo, e o sistema está fechado, temos uma barreira ocupacional. Este gráfico ajuda gestores a otimizar a escala de profissionais.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Selecionando Brasil (SUS), EUA (Privado/Misto) e um exemplo de país com desafios econômicos (ex: South Africa)
paises_sus = ['Brazil', 'United States', 'South Africa']
df_sus = df[df['country'].isin(paises_sus)]

plt.figure(figsize=(14, 7))
sns.lineplot(data=df_sus, x='date', y='people_vaccinated_per_hundred', hue='country', linewidth=2.5)

plt.title('Acesso Universal: Percentual da População com Pelo menos uma Dose', fontsize=14, pad=15)
plt.xlabel('Data', fontsize=12)
plt.ylabel('% da População Vacinada', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

A análise da curva de people_vaccinated_per_hundred demonstra como o princípio da Universalidade do SUS impacta a velocidade de adesão. Embora os Estados Unidos tenham iniciado a campanha meses antes devido ao poder de compra e fabricação de insumos, a robustez da rede capilarizada do PNI (Programa Nacional de Imunizações) no SUS permitiu ao Brasil apresentar uma inclinação de curva constante, mitigando barreiras econômicas de acesso que são evidentes em sistemas de saúde privados ou fragmentados.

## Conclusão: O Nexo entre a Ciência de Dados e a Clínica Ocupacional

A análise epidemiológica realizada neste notebook demonstra que a ciência de dados é uma ferramenta indissociável da prática clínica em saúde pública. Ao traduzirmos os códigos do Python, as manipulações com o Pandas e as visualizações gráficas geradas no ambiente do Google Colab, não estamos lidando apenas com números isolados, mas com registros matemáticos de vidas, rotinas e territórios. O pico de vacinação diária registrado nos dados reflete a força da descentralização do SUS. A engenharia de saúde pública brasileira, sustentada pela Estratégia Saúde da Família (ESF), transforma postos de saúde locais em pontos de imunização em massa de alta eficiência. Essa estrutura demonstra que o gargalo do sistema não era a capacidade de aplicação (recursos humanos e logística de ponta), mas sim o cronograma de entrega global de insumos (matéria-prima).

O atraso na inclinação da curva de cobertura vacinal brasileira no início de 2021 — decorrente de entraves políticos e logísticos na aquisição de insumos — possui um nexo de causalidade direto com o sofrimento cotidiano e a ruptura funcional da população atendida pelo SUS. 

### A Tradução dos Dados na Realidade da Terapia Ocupacional:

1. **A Linha do Gráfico vs. O Cotidiano Desestruturado:** Cada ponto abaixo da meta de imunização no gráfico de linha representou um aumento nas internações por COVID-19. Para a Terapia Ocupacional, isso se traduziu na desestruturação massiva das dinâmicas familiares, na perda de papéis ocupacionais (provedores, avós, cuidadores) e na instalação de um luto coletivo que paralisou a autonomia de milhares de sujeitos.
2. **O Impacto Ocupacional da Variante Gama:** O estouro da curva de letalidade gerado pela falta de barreiras vacinais precoces expôs a população mais vulnerável a uma escolha cruel entre o isolamento sem recursos ou o engajamento ocupacional inseguro no trabalho informal, configurando um cenário clássico de **Injustiça Ocupacional**.
3. **A Demanda por Reabilitação no SUS:** O volume de casos acumulados que o código nos mostra transformou-se na sobrecarga direta dos serviços de reabilitação. O atraso na vacina gerou uma geração de sobreviventes da intubação acometidos pela Síndrome Pós-COVID.
